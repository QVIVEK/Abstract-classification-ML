```python
import pandas as pd

# Corrected URL path and spelling ('bussiness' with double 's' and branch 'main')
csv_url = '/business_raw_20000.csv'
df = pd.read_csv(csv_url)
display(df.head())
```

Data Sourcing
Through Scopus advance Search  
for business : SUBJAREA ( BUSI ) AND ( LIMIT-TO ( DOCTYPE , "cp" ) OR LIMIT-TO ( DOCTYPE , "ar" ) ) AND ( LIMIT-TO ( SRCTYPE , "j" ) OR LIMIT-TO ( SRCTYPE , "p" ) OR LIMIT-TO ( SRCTYPE , "d" ) ) truncated to 20000 latest results (limit of scopus export)



```python
# Removing link and adding serial number, also deleting entries with no abstract, also adding one column with attribute business
import pandas as pd
import numpy as np

# Explicitly drop the 'Link' column and ignore errors if already dropped
df = df.drop(columns=['Link'], errors='ignore')

#Delete entries with no abstract
col_name = "Abstract"
df = df.dropna(subset=[col_name])
df = df[df[col_name].astype(str).str.strip() != '']

#Add a sequential Serial Number column at the beginning
if 'Serial Number' not in df.columns:
    df.insert(0, 'Serial Number', range(1, len(df) + 1))

#Add a column with the attribute 'business'
df['attribute'] = 'business'
print(df.count())

# Display the cleaned DataFrame
display(df.head())

# Export the cleaned DataFrame to a CSV file
df.to_csv('business_cleaned.csv', index=False)
```
